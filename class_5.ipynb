```python
from sklearn.preprocessing import PowerTransformer
```

```python
x=np.array([[-1],[0],[3],[5],[10],[50]])
pt = PowerTransformer(method="yeo-johnson")
x_transformed = pt.fit_transform(x)
print("Oringinal:")
print(x.ravel())
print("\nTramsformed:")
print(np.round(x_transformed.ravel(),3))
print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
y=np.array([[1],[1],[3],[5],[10],[50]])
pt = PowerTransformer(method="box-cox")
y_transformed = pt.fit_transform(y)
print("Oringinal:")
print(y.ravel())
print("\nTramsformed:")
print(np.round(y_transformed.ravel(),3))
print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Aman', 'Priya', 'Rahul', 'Sneha', 'Vikram', 'Pooja', 'Rohan'],
    # Categorical Columns (Nulls: None / np.nan)
    'Gender': ['Male', 'Female', np.nan, 'Female', 'Male', np.nan, 'Male'],
    'Department': ['IT', 'HR', 'Finance', np.nan, 'IT', 'Finance', np.nan],
    'City': ['Delhi', np.nan, 'Mumbai', 'Bangalore', 'Delhi', 'Pune', np.nan],
    # Numerical Columns (Nulls: np.nan)
    'Age': [25, 28, np.nan, 32, np.nan, 29, 35],
    'Salary': [50000, 62000, 48000, np.nan, 75000, np.nan, 80000],
    'Experience_Years': [2.0, 4.5, np.nan, 7.0, 5.0, 3.5, np.nan]
}

df = pd.DataFrame(data)

# Display dataset & null values count
print(df)
print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer=SimpleImputer(strategy='mean')
imputer.fit_transform(df[['Salary']])

# imputer = SimpleImputer(strategy='mean')
# df[['Salary']] = imputer.fit_transform(df[['Salary']])
```

```python
imputer=SimpleImputer(strategy='mean')
df['new_salary']=imputer.fit_transform(df[['Salary']]).ravel()

# df['new_salary'] = imputer.fit_transform(df[['Salary']]).ravel()
```
