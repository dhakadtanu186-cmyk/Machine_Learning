```python
import pandas as pd
import numpy as np

from statsmodels.stats.outliers_influence import (variance_inflation_factor)
df = pd.DataFrame({
    "Area_sqft": [1000, 1500, 2000, 2500, 3000, 3500],
    # "Area_sqm": [92.9, 139.4, 185.8, 232.3, 278.7, 325.2],
    # "Bedrooms": [2, 3, 3, 4, 4, 5],
    "Age": [15, 10, 8, 5, 3, 2]
})

x=df.copy()
vif_data=pd.DataFrame()
vif_data["Feature"]=x.columns

vif_data["VIf"]=[
    variance_inflation_factor(x.values,i)for i in range(x.shape[1])]
print("\nVIF Results:")
vif_data.round(2)
```
