```python
import pandas as pd 
import numpy as np 
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,max_error
```

```python
x1 = np.linspace(12,50,num=30,dtype=int)
x2 = np.linspace(22,60,num=30,dtype=int)
y = np.linspace(48,162,num=30,dtype=int)

data={
    'x1':x1,
    'x2':x2,
    'y':y}
df =pd.DataFrame(data)
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
lr=LinearRegression()
lr.fit(x,y) 
```

```python
data
```

```python
y_pred=lr.predict(x)
```

```python
y_pred
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
mean_squared_error(y,y_pred) 
```

```python
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y,y_pred)
```
